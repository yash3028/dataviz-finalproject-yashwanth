```python

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

print("=" * 70)
print("LOADING DATASETS")
print("=" * 70)

accidents = pd.read_csv("./dataset/Accident_Information.csv",low_memory=False,encoding="latin1")
vehicles = pd.read_csv("./dataset/Vehicle_Information.csv",low_memory=False,encoding="latin1")

print("Accident Information Dataset Loaded Successfully.")
print("Vehicle Information Dataset Loaded Successfully.")


print("\n" + "=" * 70)
print("DATASET SHAPES")
print("=" * 70)

print(f"Accident Dataset : {accidents.shape}")
print(f"Vehicle Dataset  : {vehicles.shape}")


print("\n" + "=" * 70)
print("ACCIDENT DATASET - FIRST FIVE RECORDS")
print("=" * 70)

display(accidents.head())

print("\n" + "=" * 70)
print("VEHICLE DATASET - FIRST FIVE RECORDS")
print("=" * 70)

display(vehicles.head())


print("\n" + "=" * 70)
print("ACCIDENT DATASET - LAST FIVE RECORDS")
print("=" * 70)

display(accidents.tail())

print("\n" + "=" * 70)
print("VEHICLE DATASET - LAST FIVE RECORDS")
print("=" * 70)

display(vehicles.tail())

print("\n" + "=" * 70)
print("ACCIDENT DATASET INFORMATION")
print("=" * 70)

accidents.info()

print("\n" + "=" * 70)
print("VEHICLE DATASET INFORMATION")
print("=" * 70)

vehicles.info()
print("\n" + "=" * 70)
print("ACCIDENT DATA TYPES")
print("=" * 70)

display(accidents.dtypes)

print("\n" + "=" * 70)
print("VEHICLE DATA TYPES")
print("=" * 70)

display(vehicles.dtypes)

print("\n" + "=" * 70)
print("ACCIDENT DATASET COLUMNS")
print("=" * 70)

for column in accidents.columns:
    print(column)

print("\n" + "=" * 70)
print("VEHICLE DATASET COLUMNS")
print("=" * 70)

for column in vehicles.columns:
    print(column)

print("\n" + "=" * 70)
print("ACCIDENT DATASET SUMMARY")
print("=" * 70)

display(accidents.describe(include="all"))

print("\n" + "=" * 70)
print("VEHICLE DATASET SUMMARY")
print("=" * 70)

display(vehicles.describe(include="all"))

print("\n" + "=" * 70)
print("MISSING VALUES - ACCIDENT DATASET")
print("=" * 70)

missing_acc = accidents.isnull().sum().sort_values(ascending=False)

display(missing_acc[missing_acc > 0])

print("\n" + "=" * 70)
print("MISSING VALUES - VEHICLE DATASET")
print("=" * 70)

missing_vehicle = vehicles.isnull().sum().sort_values(ascending=False)

display(missing_vehicle[missing_vehicle > 0])

print("\n" + "=" * 70)
print("DUPLICATE RECORDS")
print("=" * 70)

print("Accident Dataset :", accidents.duplicated().sum())

print("Vehicle Dataset  :", vehicles.duplicated().sum())

print("\n" + "=" * 70)
print("UNIQUE ACCIDENT IDs")
print("=" * 70)

print("Accident Dataset :", accidents["Accident_Index"].nunique())

print("Vehicle Dataset  :", vehicles["Accident_Index"].nunique())

print("\n" + "=" * 70)
print("MERGING DATASETS")
print("=" * 70)

merged_df = pd.merge(
    accidents,
    vehicles,
    on="Accident_Index",
    how="inner"
)
print(merged_df.shape)
display(merged_df.head())

print("Datasets merged successfully.")

print("\n" + "=" * 70)
print("MERGED DATASET SHAPE")
print("=" * 70)

print(merged_df.shape)

print("\n" + "=" * 70)
print("FIRST FIVE RECORDS OF MERGED DATASET")
print("=" * 70)

display(merged_df.head())

print("\n" + "=" * 70)
print("MERGED DATASET INFORMATION")
print("=" * 70)

merged_df.info()

print("\n" + "=" * 70)
print("DUPLICATES AFTER MERGING")
print("=" * 70)

print(merged_df.duplicated().sum())

print("\n" + "=" * 70)
print("TOP 30 COLUMNS WITH MISSING VALUES")
print("=" * 70)

display(
    merged_df.isnull()
             .sum()
             .sort_values(ascending=False)
             .head(30)
)

merged_df.to_csv(
    "./dataset/merged_accident_vehicle.csv",
    index=False
)

print("\nMerged dataset saved as")

print("./dataset/merged_accident_vehicle.csv")

print("\n" + "=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print("Total Rows    :", merged_df.shape[0])

print("Total Columns :", merged_df.shape[1])

print("Unique Accidents :", merged_df["Accident_Index"].nunique())

print("Duplicate Rows :", merged_df.duplicated().sum())

print("\nFile 1 Completed Successfully.")
```

```python

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

print("=" * 70)
print("LOADING MERGED DATASET")
print("=" * 70)

df = pd.read_csv("data/merged_accident_vehicle.csv")

print("Dataset Loaded Successfully.")
print("Shape :", df.shape)

print("\n" + "=" * 70)
print("REPLACING INVALID VALUES")
print("=" * 70)

invalid_values = [
    "NA",
    "Data missing or out of range",
    "Unknown",
    "",
    " "
]

df.replace(invalid_values, np.nan, inplace=True)

print("Invalid values replaced with NaN.")

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

missing = df.isnull().sum().sort_values(ascending=False)

display(missing[missing > 0])

print("\n" + "=" * 70)
print("REMOVING DUPLICATES")
print("=" * 70)

print("Duplicates Before :", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicates After :", df.duplicated().sum())

print("\nConverting Date...")

df["Date"] = pd.to_datetime(
    df["Date"],
    dayfirst=True,
    errors="coerce"
)

print("Converting Time...")

df["Time"] = pd.to_datetime(
    df["Time"],
    format="%H:%M",
    errors="coerce"
)

numeric_columns = [
    "Age_of_Vehicle",
    "Engine_Capacity_.CC.",
    "Speed_limit",
    "Latitude",
    "Longitude",
    "Number_of_Casualties",
    "Number_of_Vehicles",
    "Driver_IMD_Decile",
    "Vehicle_Location.Restricted_Lane"
]

for col in numeric_columns:

    if col in df.columns:

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

print("Numeric conversion completed.")

print("\nFilling Numeric Missing Values...")

num_cols = df.select_dtypes(include=np.number).columns

for col in num_cols:

    df[col].fillna(
        df[col].median(),
        inplace=True
    )
print("Filling Categorical Missing Values...")

cat_cols = df.select_dtypes(include="object").columns

for col in cat_cols:

    if df[col].mode().empty == False:

        df[col].fillna(
            df[col].mode()[0],
            inplace=True
        )
print("\nCreating New Features...")

df["Year"] = df["Date"].dt.year

df["Month"] = df["Date"].dt.month_name()

df["Month_Number"] = df["Date"].dt.month

df["Day"] = df["Date"].dt.day_name()

df["Hour"] = df["Time"].dt.hour

df["Weekend"] = df["Day"].isin(
    ["Saturday","Sunday"]
)

def get_period(hour):

    if hour < 6:
        return "Night"

    elif hour < 12:
        return "Morning"

    elif hour < 17:
        return "Afternoon"

    elif hour < 21:
        return "Evening"

    else:
        return "Night"

df["Time_Period"] = df["Hour"].apply(get_period)

df["Vehicle_Age_Group"] = pd.cut(

    df["Age_of_Vehicle"],

    bins=[-1,5,10,20,100],

    labels=[
        "0-5 Years",
        "6-10 Years",
        "11-20 Years",
        "20+ Years"
    ]

)

df["Engine_Group"] = pd.cut(

    df["Engine_Capacity_.CC."],

    bins=[0,1000,1600,2500,10000],

    labels=[
        "<1000 cc",
        "1000-1600 cc",
        "1600-2500 cc",
        ">2500 cc"
    ]

)

severity_map = {
    "Slight":1,
    "Serious":2,
    "Fatal":3
}

df["Severity_Code"] = df["Accident_Severity"].map(severity_map)

print("\n" + "=" * 70)
print("MISSING VALUES AFTER CLEANING")
print("=" * 70)

display(

    df.isnull()
      .sum()
      .sort_values(ascending=False)
      .head(30)

)

print("\nDataset Information")

df.info()

print("\nFinal Shape")

print(df.shape)

display(df.head())
df.to_csv(

    "data/cleaned_accidents.csv",

    index=False

)

print("\nCleaned dataset saved successfully.")

print("Location : data/cleaned_accidents.csv")

print("\n" + "=" * 70)
print("DATA CLEANING SUMMARY")
print("=" * 70)

print("Rows :", df.shape[0])

print("Columns :", df.shape[1])

print("Duplicate Rows :", df.duplicated().sum())

print("Remaining Missing Values :", df.isnull().sum().sum())

print("\nFile 2 Completed Successfully.")
```
